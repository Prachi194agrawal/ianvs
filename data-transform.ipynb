```python
from datasets import load_dataset

dataset = load_dataset("jeggers/gpqa_formatted", "diamond")["train"]

dataset
```

```python
prompt = """{input}

A) {A}
B) {B}
C) {C}
D) {D}
"""
```

```python
import json
ianvs_dataset = []

for item in dataset:
    choices = item["options"]
    system_prompt = "Answer the following multiple choice question. The last line of your response should be of the following format: 'ANSWER: $INDEX$' where INDEX is one of A, B, C, or D. Think step by step before answering.\n"
    query = prompt.format(input=item["Question"], A=choices[0], B=choices[1], C=choices[2], D=choices[3])
    answer = item["answer"]
    level_1_dim = "single-modal"
    level_2_dim = "text"
    level_3_dim = "knowledge Q&A"
    level_4_dim = "Uncategorized"

    ianvs_dataset.append({
        "prompt": system_prompt,
        "query": query,
        "response": ["A", "B", "C", "D"][answer],
        "level_1_dim": level_1_dim,
        "level_2_dim": level_2_dim,
        "level_3_dim": level_3_dim,
        "level_4_dim": level_4_dim
    })

with open("data-full.jsonl", "w") as f:
    for item in ianvs_dataset:
        f.write(json.dumps(item) + "\n")
```
